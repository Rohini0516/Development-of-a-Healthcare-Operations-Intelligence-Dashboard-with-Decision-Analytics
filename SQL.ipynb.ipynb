SQL Coding Interview Questions for Freshers
To begin, we'll set up an in-memory SQLite database using sqlite3 in Python. This will allow us to create tables and insert data directly within the Colab environment, and then run SQL queries against it.

```python
import sqlite3
import pandas as pd

# Connect to an in-memory SQLite database
conn = sqlite3.connect(":memory:")
c = conn.cursor()

# Create Employees table
c.execute("""
CREATE TABLE Employees (
    EmpID INTEGER PRIMARY KEY,
    EmpName TEXT NOT NULL,
    DepartmentID INTEGER,
    Salary INTEGER
);
""")

# Create Departments table
c.execute("""
CREATE TABLE Departments (
    DepartmentID INTEGER PRIMARY KEY,
    DepartmentName TEXT NOT NULL
);
""")

# Insert data into Employees
employees = [
    (1, 'Alice', 101, 60000),
    (2, 'Bob', 102, 80000),
    (3, 'Charlie', 101, 65000),
    (4, 'David', 102, 90000),
    (5, 'Eve', 101, 70000),
    (6, 'Frank', 103, 75000)
]

c.executemany("""
INSERT INTO Employees
(EmpID, EmpName, DepartmentID, Salary)
VALUES (?, ?, ?, ?);
""", employees)

# Insert data into Departments
departments = [
    (101, 'HR'),
    (102, 'Engineering'),
    (103, 'Marketing')
]

c.executemany("""
INSERT INTO Departments
(DepartmentID, DepartmentName)
VALUES (?, ?);
""", departments)

# Commit changes
conn.commit()

print("Database and tables created with sample data.")
```

Question 11: HAVING

Problem Statement: Display only departments that have more than two employees.

```python
query11 = """
SELECT
    DepartmentID,
    COUNT(*) AS TotalEmployees
FROM Employees
GROUP BY DepartmentID
HAVING COUNT(*) > 2;
"""

display(run_sql_query(query11))
```
