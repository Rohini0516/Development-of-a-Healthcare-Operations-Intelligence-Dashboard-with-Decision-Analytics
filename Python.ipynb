## Question: Reverse a string
# Problem Statement  
Write a Python function to reverse a given string.

# Example  
Input: s = "hello"  
Output: olleh

# Explanation  
We can reverse a string by iterating through it from the end to the beginning and appending each character to a new string, or by using Python’s slicing feature.

```python
# Function to reverse a string
def reverse_string(s):
    reversed_s = ""
    for i in range(len(s) - 1, -1, -1):
        reversed_s += s[i]
    return reversed_s

# Example usage
input_string = "hello"
output_string = reverse_string(input_string)
print(f"Original string: {input_string}")
print(f"Reversed string: {output_string}")

```

# Question : Check whether a number is prime

# Problem Statement  
Write a Python function that takes an integer as input and returns True if the number is prime, and False otherwise.

# Example  
Input: n = 7 → Output: True  
Input: n = 10 → Output: False

# Explanation  
A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself. We can check for primality by iterating from 2 up to the square root of the number and checking for divisibility. If any number in this range divides n evenly, then n is not prime.

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

print(is_prime(7))   # True
print(is_prime(10))  # False
```

# Question: Find the largest number in a list

# Problem Statement  
Write a Python function to find the largest number in a given list of numbers.

# Example  
Input: numbers = [3, 1, 9, 4, 2]  
Output: 9

# Explanation  
We can find the largest number by initializing a variable with the first element of the list and then iterating through the rest of the list, updating the variable whenever a larger number is found.

```python
def find_largest_number(numbers):
    if not numbers:
        return None
    largest = numbers[0]
    for num in numbers[1:]:
        if num > largest:
            largest = num
    return largest

print(find_largest_number([3, 1, 9, 4, 2]))
```

# Question: Check whether a string is a palindrome

# Problem Statement  
Write a Python function that checks if a given string is a palindrome. A palindrome reads the same forwards and backwards.

# Example  
Input: "madam" → Output: True  
Input: "hello" → Output: False

# Explanation  
We can check if a string is a palindrome by comparing the original string with its reversed version. If they are the same, the string is a palindrome.

```python
def is_palindrome(s):
    return s == s[::-1]

print(is_palindrome("madam"))  # True
print(is_palindrome("hello"))  # False
```

# Question:Count vowels in a string

# Problem Statement  
Write a Python function to count the number of vowels (a, e, i, o, u) in a given string. The function should be case-insensitive.

# Example
Input: s = "Hello World"  
Output: 3 (e, o, o)

# Explanation  
We can iterate through each character of the string, convert it to lowercase, and check if it is present in a predefined set of vowels. If it is, we increment a counter.

```python
def count_vowels(s):
    vowels = "aeiou"
    return sum(1 for ch in s.lower() if ch in vowels)

print(count_vowels("Hello World"))
```

# Question: Fibonacci Series

# Problem Statement  
Write a Python function to generate the Fibonacci series up to n terms. The Fibonacci sequence is a series where the next term is the sum of the previous two terms. The first two terms are 0 and 1.

# Example  
Input: n = 7  
Output: [0, 1, 1, 2, 3, 5, 8]

# Explanation  
We start with the first two terms (0 and 1) and then use a loop to generate subsequent terms by adding the previous two. We store these terms in a list.

```python
# Function to generate Fibonacci series up to n terms
def fibonacci_series(n):
    fib_list = []
    if n <= 0:
        return fib_list
    elif n == 1:
        return [0]
    else:
        a, b = 0, 1
        fib_list.extend([a, b])
        for _ in range(2, n):
            next_fib = a + b
            fib_list.append(next_fib)
            a, b = b, next_fib
        return fib_list

# Example usage
print(fibonacci_series(7))  # [0, 1, 1, 2, 3, 5, 8]
```

# Question: Maximum Subarray Sum (Kadane’s Algorithm)

# Problem Statement  
Write a Python function to find the maximum sum of a contiguous subarray within a given list of integers.

# Example  
Input: arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]  
Output: 6 (subarray [4, -1, 2, 1])

# Explanation  
We use Kadane’s Algorithm: iterate through the array, at each step decide whether to add the current element to the existing sum or start a new subarray. Keep track of the maximum sum seen so far.

```python
# Function to find maximum subarray sum
def max_subarray_sum(arr):
    max_sum = arr[0]
    current_sum = arr[0]
    for num in arr[1:]:
        current_sum = max(num, current_sum + num)
        max_sum = max(max_sum, current_sum)
    return max_sum

# Example usage
print(max_subarray_sum([-2,1,-3,4,-1,2,1,-5,4]))  # 6
```

# Question: Factorial of a number

# Problem Statement  
Write a Python function to calculate the factorial of a given number.

# Example  
Input: n = 5  
Output: 120

# Explanation  
Factorial of a number n is the product of all positive integers less than or equal to n. For example, 5! = 5 × 4 × 3 × 2 × 1.


```python
# Function to calculate factorial
def factorial(n):
    if n < 0:
        return None  # Factorial not defined for negative numbers
    result = 1
    for i in range(1, n+1):
        result *= i
    return result

# Example usage
print(factorial(5))  # 120
print(factorial(0))  # 1
```

# Question: Find the second largest number in a list
# Problem Statement  
Write a Python function to find the second largest number in a given list.

# Example  
Input: [5, 2, 8, 1, 9]  
Output: 8

# Explanation  
We can remove duplicates, sort the list in descending order, and return the second element.

```python
# Function to find the second largest number
def second_largest(lst):
    unique = sorted(set(lst), reverse=True)
    return unique[1] if len(unique) >= 2 else None

# Example usage
print(second_largest([5, 2, 8, 1, 9]))  # 8
print(second_largest([10, 10, 10]))     # None
```
